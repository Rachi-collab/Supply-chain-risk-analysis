## Problem statement

Modern supply chains generate vast amounts of data, yet organizations often operate reactively rather than proactively when managing disruptions. The core challenge is to synthesize historical logistics, supplier performance, and procurement data to identify hidden vulnerabilities, predict high-risk failure points, and build an actionable framework that optimizes inventory and transit resilience.

## Proposed solution

This project transforms reactive logistics into a proactive strategy by building an analytical pipeline:

Data Integration: Aggregates procurement and shipping data to engineer risk metrics (e.g., Supplier Reliability Index).

Predictive Modeling: Trains Machine Learning models (Random Forest/XGBoost) to forecast shipment delays and stockout probabilities.

Prescriptive Action: Deploys a dynamic safety stock framework alongside an interactive dashboard to monitor high-risk transit routes and optimize inventory buffers.
